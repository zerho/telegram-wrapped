```python
fmt = '%d/%m/%y, %H:%M:%S'

firstDay = pd.to_datetime(df_wa['date'].min(), format=fmt)
lastDay = pd.to_datetime(df_wa['date'].max(), format=fmt)


print("In this chat there are:")

print(f"Date of first message: {firstDay}")
print(f"{lastDay - firstDay} days")
print(f"{df_wa['sender'].count()} messages")
print(f"{df_wa['sender'].nunique()} members")

# Count di tutti i messaggi per singolo sender
df_wa['sender'].value_counts()
```
